```python
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash")
```

```python
class blogState(TypedDict):
    title: str
    outline: str
    blog: str
```

```python
def llm_outline(state: blogState) -> blogState:
    title = state["title"]
    prompt = f"Generate an outline for a blog post with the title: {title}"
    outline = model.invoke(prompt).content
    state["outline"] = outline
    return state
```

```python
def llm_blog(state: blogState) -> blogState:
    outline = state["outline"]
    prompt = f"Generate a blog post based on the following outline: {outline}"
    blog = model.invoke(prompt).content
    state["blog"] = blog
    return state
```

```python
graph = StateGraph(blogState)

graph.add_node("llm_outline", llm_outline)
graph.add_node("llm_blog", llm_blog)

graph.add_edge(START, "llm_outline")    
graph.add_edge("llm_outline", "llm_blog")
graph.add_edge("llm_blog", END)

workflow =graph.compile()
```

```python
initial_state = {"title": "The Future of AI in Healthcare"}
final_state = workflow.invoke(initial_state)
print(final_state)
```

```python
print(final_state["blog"])  # Print the generated blog post
```
